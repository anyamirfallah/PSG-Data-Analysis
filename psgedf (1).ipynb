```python
raw = mne.io.read_raw_edf( r"D:\hospital\New folder\1\edf\1.edf",
    preload=True
)

raw.plot(
    duration=30,
    n_channels=10,
    scalings="auto"
)
```

```python
print("Number of channels:", len(raw.ch_names))
print("\nChannel names:")

for i, ch in enumerate(raw.ch_names):
    print(i, ":", ch)

print("\nSampling frequency:", raw.info["sfreq"], "Hz")
print("Duration:", raw.times[-1], "seconds")
```

```python
for i, ch in enumerate(raw.info["chs"]):
    print(
        i,
        "|",
        ch["ch_name"],
        "| unit:", ch["unit"],
        "| kind:", ch["kind"]
    )
```

```python
import numpy as np

data = raw.get_data()

for i, ch in enumerate(raw.ch_names):
    signal = data[i]

    print(
        f"{i:2d} | {ch:15s} | "
        f"min={signal.min():.6e} | "
        f"max={signal.max():.6e} | "
        f"mean={signal.mean():.6e} | "
        f"std={signal.std():.6e}"
    )
```

```python
channels = [
    "F4:A1",
    "C4:A1",
    "O2:A1",
    "EOG1:A2",
    "EMG",
    "Flow DVH"
]

raw.plot(
    picks=channels,
    duration=30,
    start=0,
    scalings="auto"
)
```
